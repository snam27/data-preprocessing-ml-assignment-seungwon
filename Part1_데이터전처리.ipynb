### 요구사항

1. 원본 데이터를 보존한 상태에서 정제용 데이터를 생성하세요.

### 힌트

- 변환할 수 없는 문자열은 결측치로 바꾼 뒤 일관되게 처리할 수 있습니다.
- 이상치는 무조건 삭제하기보다 실제 우수 고객의 행동일 가능성도 고려하세요.
- IQR은 이상치 후보를 확인하는 대표적인 방법입니다.

```python
# TODO: 원본을 보존하고 정제용 데이터 df_clean을 생성하세요.
df_clean = df.copy()
df_clean

# TODO: 수치형·날짜형 자료형과 범주 표기를 정리하세요.
## 데이터 전처리
# 대소문자 표기통일 및 텍스트 앞뒤 공백 제거
target_cols = ['Gender', 'Region', 'MembershipLevel', 'CouponUsed', 'PreferredCategory']

for col in target_cols:
  df_clean[col] = df_clean[col].astype(str).str.strip().str.title()  # str.strip(): 앞뒤공백제거 / str.title(): 앞대문자, 뒤소문자 

# 각 컬럼의 고유값 확인
'''for col in target_cols:
  print(f"=== {col} ===")
  print(df_clean[col].unique())
  print()
'''

# 약어 없애기
# Gender 칼럼 
df_clean['Gender'] = df_clean['Gender'].replace({'F': 'Female', 'M': 'Male'}) # Ohter는 일단 남겨둠
df_clean['Gender'].unique()
# CouponUsed 칼럼
df_clean['CouponUsed'] = df_clean['CouponUsed'].replace({'Y': 'Yes', 'N': 'No'})
df_clean['CouponUsed'].unique()

# Age 칼럼 별개로 처리
# '세' 제거
df_clean['Age'] = df_clean['Age'].astype(str).str.replace('세', '').str.strip()
# 'thirty'를 '30'으로 변경
df_clean['Age'] = df_clean['Age'].replace('thirty', '30')

## 자료형 정리
# 수치형으로 변환
df_clean['Age'] = pd.to_numeric(df_clean['Age'], errors='coerce')  # 나머지 결측치 문자열은 nan처리
df_clean['AveragePurchaseAmount'] = pd.to_numeric(df_clean['AveragePurchaseAmount'], errors='coerce') # nan값 남겨둠
#  CouponUsed 수치형으로 변환할지 고려해보기 #

# 날짜형으로 변환
df_clean['LastPurchaseDate'] = pd.to_datetime(df_clean['LastPurchaseDate'], errors='coerce')
df_clean['SignupDate'] = pd.to_datetime(df_clean['SignupDate'], errors='coerce')

## 결과
# 자료형 확인
df_clean.info()

# 고유값 확인
# 모든 컬럼의 고유값 
#for col in df_clean.columns:
#    print(f"[{col}] : {df_clean[col].unique()}")
```

```python
# TODO: 결측치와 전체 행 기준 중복 데이터를 처리하세요.
## 결측치 처리

## 범주형 칼럼
# 범주형 칼럼
categorical_cols = [
    "Gender",
    "Region",
    "MembershipLevel",
    "PreferredCategory"
]

# 전처리중 Nan으로 변경된 문자열 -> 다시 NaN값으로 복원
df_clean[categorical_cols] = df_clean[categorical_cols].replace(
    ["Nan"],np.nan
)

# 결측치를 Unknown으로 처리
df_clean[categorical_cols] = df_clean[categorical_cols].fillna("Unknown")


## 수치형 칼럼
# Age 칼럼: 결측치를 중앙값(median)으로 채운후 int로 변환
median_age = df_clean['Age'].median()
df_clean['Age'] = df_clean['Age'].fillna(median_age).astype(int)  # median: 39 == mode: 39

# AveragePurchaseAmount 칼럼: 결측치를 중앙값으로 채움
df_clean["AveragePurchaseAmount"] = df_clean["AveragePurchaseAmount"].fillna(
    df_clean["AveragePurchaseAmount"].median()   # median: 117500.0
)

# SatisfactionScore 칼럼: 결측치를 최빈값(mode)로 채움
df_clean["SatisfactionScore"] = df_clean["SatisfactionScore"].fillna(
    df_clean["SatisfactionScore"].median()  # mode = 3.6 (min: 1.1 / max: 10)
)

## 날짜형 칼럼 -> 이건 가짜 날짜를 생성하는 것보다는 결측치 일단 놔두기로 함.
# LastPurchaseDate 
df_clean["LastPurchaseDate"].isna().sum() # 결측치: 65개

# SignupDate       
df_clean["SignupDate"].isna().sum()  # 결측치: 15개


## 중복치 제거
# 전체 행 기준 중복치는 제거
df_clean = df_clean.drop_duplicates()

# TODO: 처리 전후 결과를 확인하세요.
print("===== 처리 전 =====")
print("행 수:", len(df))
print("결측치 수:", df.isna().sum().sum())
print("중복 행 수:", df.duplicated().sum())

display(df.head())


print("\n===== 처리 후 =====")
print("행 수:", len(df_clean))
print("결측치 수:", df_clean.isna().sum().sum())  # 날짜형 결측치 수( 64 + 15 = 79개) 남아있음
print("중복 행 수:", df_clean.duplicated().sum())

display(df_clean.head())
```

```python
def get_iqr_bounds(series):
    # TODO: Q1, Q3, IQR, 하한과 상한을 반환하세요.
    # Q1, Q3 계산
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)

    # IQR 계산
    IQR = Q3 - Q1

    # 이상치 하한과 상한
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    return Q1, Q3, IQR, lower_bound, upper_bound
    

# TODO: 주요 수치형 컬럼의 이상치를 탐지하고 처리하세요.

# 이상치 탐지
numeric_cols = ["Age", "VisitCount", "TotalPurchaseAmount", "AveragePurchaseAmount"]

print("=== 처리 전 이상치 확인 ===")

for col in numeric_cols:
    Q1, Q3, IQR, lower_bound, upper_bound = get_iqr_bounds(df_clean[col])

    # 이상치 탐지
    outliers = df_clean[
        (df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)
    ]
   
    print(f"{col}")
    print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
    print(f"하한: {lower_bound}, 상한: {upper_bound}")
    print(f"이상치 개수: {len(outliers)}")
    print()

# 컬럼별로 음수값이 있는지 확인
print("=== 음수 값 확인 ===")

for col in numeric_cols:
    print(f"{col}: 음수 {((df_clean[col] < 0).sum())}개")

## 수치형 컬럼 4개 모두 꼭 이상치 처리가 필요한지 잘 모르겠음(나이, 방문횟수, 총구매금액, 평균구매금액).
## 고령이 있을 수도 있고, 횟수가 실제로 많을 수도 있고, 실제로 고액구매가 이뤄져서 평균구매 금액 또한 높아졌을 수도 있다.
## 따라서 아래의 이상치 처리(iqr 경계값으로 outlier들을 대체하는 방법)를 시행하지 않기로 결정함.

'''
# 이상치 처리: IQR 범위 밖의 값을 경계값으로 대체 
for col in numeric_cols:
    Q1, Q3, IQR, lower_bound, upper_bound = get_iqr_bounds(df_clean[col])

    df_clean[col] = df_clean[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )


# 처리 전후 이상치 상태 확인
print("=== 처리 후 이상치 확인 ===")

for col in numeric_cols:
    Q1, Q3, IQR, lower_bound, upper_bound = get_iqr_bounds(df_clean[col])

    outliers = df_clean[
        (df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)
    ]

    print(f"{col}: 이상치 개수 = {len(outliers)}")

'''
```

### 4-2. 결측치·이상치 처리 방법의 선택 근거

- 결측치 처리 방법:

수치형 변수는 변수의 특성과 분포를 고려하여 중앙값 또는 최빈값으로 결측치를 대체하고, 범주형 변수는 결측치를 unknown으로 처리하였다. 날짜형 변수의 경우 임의의 날짜를 입력하기보다 결측 상태(NaT)를 유지하는 방법을 적용하였다.

- 해당 방법을 선택한 이유:

수치형 변수에 평균을 사용할 경우 이상치의 영향을 받을 수 있기 때문에 상대적으로 극단값의 영향을 덜 받는 중앙값이나 최빈값을 활용하는 것이 적절하다고 판단하였다. 

범주형 변수는 임의의 값을 넣는 것(예: gender칼럼의 결측치에 Male, Female을 넣는 것보단)보단 결측치를 알수없음으로 표기해 따로 처리하는게 더 적합하다고 생각했다.

날짜형 변수는 임의의 값을 입력하면 실제 날짜로 오해될 수 있으므로, 날짜 정보가 없는 경우 NaT로 유지하는 것이 데이터의 의미를 보존하는 데 적절하다고 판단했다.

- 이상치 처리 방법:

수치형 변수의 이상치는 IQR(사분위범위)을 기준으로 하여 하한값과 상한값을 확인하고, 데이터의 특성을 고려하여 필요한 경우 범위를 벗어난 값을 조정하려고 하였으나, 

수치형 변수들을 보니 나이, 방문횟수, 표준/총 구매금액 모두 이상치가 실질적인 의미를 가진 데이터(예: 실제로 고령, 실제로 자주방문, 실제 고액구매)일수 있겠다고 판단하여 이상치 처리를 

시행하지는 않고 코드를 주석화 해놓았다. 

- 이상치 처리 시 주의점:

IQR 기준을 벗어났다고 해서 반드시 오류 데이터인 것은 아니다. 

특히 구매금액이나 방문횟수처럼 실제로 고객 간 차이가 크게 발생할 수 있는 변수에서는 극단값 자체가 중요한 고객 특성을 나타낼 수 있으므로, 단순히 이상치를 삭제하지 않고 실제 오류 여부와 분석 목적을 함께 고려해야 한다.

### 4-3. 파생변수의 활용 의도와 한계

- `AgeGroup` 생성 기준과 활용 목적:

Age를 연령 구간별로 구분하여 10대 이하, 10대, 20대, 30대, 40대, 50대, 60대 이상으로 그룹화하였다(데이터에 10대는 없었으므로). 

이를 통해 개별 연령을 사용하는 것보다 연령대별 고객 수, 구매금액, 방문횟수, 만족도 등의 차이를 비교하여 고객 특성을 파악할 수 있다.

- `PurchaseGrade` 생성 기준과 활용 목적:

TotalPurchaseAmount를 사분위수 기준으로 4개 그룹으로 나누어 Low, Medium, High, VIP로 분류하였다. 

이를 활용하면 구매금액 수준에 따른 고객 특성을 비교하고, 각 구매등급별 연령대, 방문횟수, 만족도, 선호 카테고리 등의 차이를 분석할 수 있다.

- 파생변수 사용 시 한계:

연속형 변수를 구간으로 나누면 원래 가지고 있던 세부적인 정보가 손실될 수 있다. 

예를 들어 같은 AgeGroup에 속한 고객이라도 실제 연령에는 차이가 있으며, PurchaseGrade 역시 동일한 등급 안에서 구매금액 차이가 존재할 수 있다. 

또한 PurchaseGrade는 현재 데이터의 사분위수를 기준으로 생성했기 때문에 데이터의 분포가 변경되면 각 등급의 기준도 달라질 수 있다. 

따라서 파생변수는 고객 특성을 쉽게 비교하기 위한 보조적인 변수로 활용하고, 필요에 따라 원래의 연속형 변수도 함께 확인하는 것이 필요할 수도 있다.
